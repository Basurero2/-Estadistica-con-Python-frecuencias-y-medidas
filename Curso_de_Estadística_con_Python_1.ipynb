### Paso 1 - Difiniendo el número de clases

#### Regla de Sturges

# $$k = 1 + \frac {10}{3}\log_{10}n$$

Indica la cantidad de clases de debemos tener para distribuir frencuencia de nuestros datos
donde:
n : cantidad de datos (filas)
k : cantidad de intervals, clases, donde se distribuira (resultado)

```python
n = datos.shape[0]    #cantidad de registros de nuestra base de datos
n
```

```python
k = 1 + (10/3)*np.log10(n)
k
#cantidad de clases, intervalos para nuestra tabla de frecuencias (bins)
```

### Paso 2 - Crear la tabla de frecuencias

```python
frecuencia_ingreso_sturges = pd.cut(x = datos.Ingreso,
                            bins = 17,
                            include_lowest = True).value_counts()
frecuencia_ingreso_sturges
#muestra la distribucion de datos para cada intervalo (no hay labels personalizados)
#labels : representacion de esos intervalos, si no hay, solo se ven los intervalos
```

```python
porcentaje_frecuencia_ingreso_sturges = pd.cut(x = datos.Ingreso,
                                               bins=17,
                                               include_lowest=True).value_counts(normalize=True).round(4)*100
porcentaje_frecuencia_ingreso_sturges
```

```python
df_distribucion_ingreso_sturges = pd.DataFrame({
    "frecuencia": frecuencia_ingreso_sturges,
    "procentaje %": porcentaje_frecuencia_ingreso_sturges
})
df_distribucion_ingreso_sturges
```

## <font color=#56B4E9>3.2 Mediana</font>
***

Para obtener la mediana de un conjunto de datos, debemos proceder de la siguiente manera:
1. Ordenar el conjunto de datos;
2. Identificar el número de observaciones (registros) del conjunto de datos ($n$);
3. Identicar el elemento del medio:

> Cuando $n$ sea impar, la posición del elemento del medio se obtendrá de la siguiente manera:


# $$Elemento_{Md} = \frac{n+1}{2}$$

> Cuando $n$ sea par, la posición del elemento mediano se obtendrá de la siguiente manera:


# $$Elemento_{Md} = \frac{n}2$$

4. Obtener la mediana:

> Cuando $n$ sea impar:


# $$Md = X_{Elemento_{Md}}$$

> Cuando $n$ sea par:


# $$Md = \frac{X_{Elemento_{Md}} + X_{Elemento_{Md}+1}}2$$
***

## <font color=#56B4E9>4.2 Box-plot</font>
***

El *box plot* da una idea de la posición, dispersión, asimetría, colas y valores extremos. La posición central está dada por la mediana y la dispersión por $IIQ$ (distancia inter cuartil). Las posiciones relativas de $Q1$, $Mediana$ y $Q3$ dan una idea de la simetría de la distribución. Las longitudes de las colas están dadas por las líneas que van desde el rectángulo a los valores remotos y por los valores extremos.

Explicacion de la grafica :
- Límite Inferior Teórico : $Q1 - 1.5 \times IIQ$
-->  (si el valor minimo de los datos es mayor o igual a eso entonces la linea se dibuja en el valor minino, pero si hay valores menores a este limite, se dibujaran como outliers estos datos, y el limite se dibujara en el valor calculado)
- Límite Superior Teórico: $Q3 + 1.5 \times IIQ$ (lo mismo, si hay valores mayores a este, se dibujaran como outliers, y el limite se dibujara en el valor calculado, sino se dibujara en el valor maximo de los datos)
- $IIQ$ : Rango Intercuartílico,  es simplemente la altura de la caja ($Q3 - Q1$), es la diferencia de valores que toma el quartil 3 con el 1, tambien es el 50% central de los datos..
- Linea inferior de la caja : quartil 1
- linea medio de la caja : quartil 2, mediana
- linea superior de la caja : quartil 3

Usaremos una libreria para calcular la  Median Abs. Deviation (MAD)
- en vez de utilizar la media, para la desviacion de cada dato, se usa la mediana, para que nuestra desviacion sea con valores concentrados y no le afecte tanto los valores extremos o muy dispersos.

Pasos
1. calcula la mediana de datos
2. Resta esa mediana a cada nota y le aplica valor absoluto: $|dato - \text{mediana}|$.
3. Calcula la mediana de esos nuevos resultados (variaciones).

- La dispersión del 50% de los casos más típicos.

## <font color=#56B4E9>5.2 Varianza</font>
***

### Varianza

La varianza se construye a partir de las diferencias entre cada observación y la media de los datos, es decir, la desviación alrededor de la media. Al calcular la varianza, las desviaciones alrededor de la media son elevadas al cuadrado.

$resumen$ : mide que tan alejados estan los datos respecto a su media, pero magnifica mas los valores extremos, dando una amplitud mas amplia, que indicaria si los datos tienen picos altos o bajos

$Usos$ :
- La varianza tiene propiedades matemáticas que el valor absoluto (MD) no tiene:Derivabilidad (Cálculo 2): En matemáticas, la función "cuadrado" ($(x)^2$) es suave y fácil de derivar en cualquier punto. La función "valor absoluto" ($|x|$) tiene una "punta" en el cero donde no se puede derivar. Esto hace que la varianza sea la herramienta preferida para optimizar modelos (como las regresiones lineales que verás más adelante).
- Aditividad: Si tienes dos grupos independientes, la varianza del total es simplemente la suma de las varianzas. Con el MD, esto no se cumple de forma tan sencilla.
- Teorema del Límite Central: La varianza es el parámetro fundamental para definir la Distribución Normal (la campana de Gauss). Casi toda la estadística inferencial (predecir el futuro con datos) se basa en la varianza.
- Descomposición (ANOVA): Puedes "partir" la varianza para saber cuánto de la variación de María se debe a sus profesores, cuánto a sus horas de estudio y cuánto al azar.

**Observacion**
- ddof : Delta Degrees of Freedom, para la divicion, 1/(n-ddof)
- pandas usa varianza de la muestra ( var(ddof=1) )
- numpy usa varianza de la poblacion ( var(ddof=0) )

### Varianza de la población
resumen : cuando se usa todo los datos

## <font color=#56B4E9>5.3 Desviación estándar</font>
***

Una de las restricciones de la varianza es el hecho de que proporciona medidas cuadráticas de las unidades originales; la varianza de las medidas de longitud, por ejemplo, está en unidades de área. Por lo tanto, el hecho de que las unidades sean diferentes dificulta la comparación de la dispersión con las variables que la definen. Una forma de eliminar esta dificultad es considerar su raíz cuadrada.

**resumen** : se usa para evitar tener diferentes unidades al usar la varianza, por ejemplo dde metros pasa a Area, y puede afectar a la interpretacion de los datos, por eso se pasa a su unidad inicial, dandole raiz cuadrada, de grado 2 a grado 1

### Desviación estándar de la población

# $$\sigma = \sqrt{\frac 1n\sum_{i=1}^{n}(X_i-\mu)^2} \Longrightarrow \sigma = \sqrt{\sigma^2}$$


### Desviación estándar de la muestra

# $$S = \sqrt{\frac 1{n-1}\sum_{i=1}^{n}(X_i-\bar{X})^2} \Longrightarrow S = \sqrt{S^2}$$
